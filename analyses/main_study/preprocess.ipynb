```python
import pandas as pd
import glob
import os

PILOT   = 'pilot_a_internal'
RAW_DIR = f'../../data/main_study/raw/{PILOT}/'
OUT_DIR = f'../../data/main_study/processed/{PILOT}/'
os.makedirs(OUT_DIR, exist_ok=True)
```

```python
# load all trial files (prefer full over partial)
full_trial_files   = glob.glob(os.path.join(RAW_DIR, '*_trials.csv'))
# exclude partials if a full exists for that participant
full_trial_files   = [f for f in full_trial_files if '_partial_' not in f]
partial_trial_files = glob.glob(os.path.join(RAW_DIR, '*_partial_trials.csv'))

# get participant IDs that have full files
full_pids = set()
for f in full_trial_files:
    pid = os.path.basename(f).replace('_trials.csv', '')
    full_pids.add(pid)

# only use partial if no full exists
partial_to_use = [f for f in partial_trial_files
                  if os.path.basename(f).replace('_partial_trials.csv', '') not in full_pids]

all_trial_files = full_trial_files + partial_to_use
trials = pd.concat([pd.read_csv(f) for f in all_trial_files], ignore_index=True)

print(f'{len(all_trial_files)} trial files loaded ({len(full_trial_files)} full, {len(partial_to_use)} partial-only)')
print(f'{len(trials)} total rows, {trials["participantID"].nunique()} participants')
trials.head(3)
```

```python
# load demographics (prefer full over partial)
full_demo_files    = [f for f in glob.glob(os.path.join(RAW_DIR, '*_demographics.csv'))
                      if '_partial_' not in f]
partial_demo_files = glob.glob(os.path.join(RAW_DIR, '*_partial_demographics.csv'))

full_demo_pids = set()
for f in full_demo_files:
    pid = os.path.basename(f).replace('_demographics.csv', '')
    full_demo_pids.add(pid)

partial_demo_to_use = [f for f in partial_demo_files
                       if os.path.basename(f).replace('_partial_demographics.csv', '') not in full_demo_pids]

all_demo_files = full_demo_files + partial_demo_to_use
demo = pd.concat([pd.read_csv(f) for f in all_demo_files], ignore_index=True)

print(f'{len(all_demo_files)} demo files loaded ({len(full_demo_files)} full, {len(partial_demo_to_use)} partial-only)')
print(f'{len(demo)} participants')
demo.head(3)
```

```python
# save processed
trials.to_csv(os.path.join(OUT_DIR, 'trials.csv'), index=False)
demo.to_csv(os.path.join(OUT_DIR, 'demographics.csv'), index=False)

print(f'saved to {OUT_DIR}')
print(f'  trials.csv:       {len(trials)} rows, {trials["participantID"].nunique()} participants')
print(f'  demographics.csv: {len(demo)} rows')
print(f'\ntrials cols: {list(trials.columns)}')
print(f'demo cols:   {list(demo.columns)}')
```
